```python
##Changing Datatypes of Columns

df['budget'] = df['budget'].astype('Int64')
df['gross'] = df['gross'].astype('Int64')
```

```python
#Comparing the year columns 
df[['year','released','released_year']]
```

```python
#r'(\d{4}) It is a Regex Function which will exactly extract the year from the released column
df['released_year'] = df['released'].str.extract((r'(\d{4})'))
```

```python
df['released_year']= df['released_year'].astype('Int64')
```

```python
#Comparing again
mismatch = df[df['year'] != df['released_year']]
print(mismatch[['name', 'year', 'released', 'released_year']].head(10))
```

```python
#year column shows the correct year , So droping the other two columns
df.drop(columns=['released','released_year'],axis=1,inplace=True)
```

```python
#renaming
df.rename(columns={'year':'released_year'})
```

```python
#Sorting the columns according to our need
df = df.sort_values(by='gross',inplace=False,ascending=False)
```

```python
top_genres = df['genre'].value_counts().nlargest(10)
x = pd.DataFrame(top_genres)
x.rename(columns={'count': 'Counts'}, inplace=True)
x
```

```python
df['profit'] = df['gross']  - df['budget']
```

```python
df['profit%'] = (df['profit'] / df['budget']) * 100  
```

```python
from datetime import datetime
df['Movie Age'] = datetime.now().year - df['year']
```

```python
plt.figure(figsize=(10,5))
ax = sns.barplot(x='genre', y='Counts', data=x,palette='flare', hue='genre')
plt.title("Top 10 genre by Movie count")
plt.tight_layout()
for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge', padding=2, fontsize=10, color='black')
plt.savefig("top_genres.png", dpi=300, bbox_inches='tight')
```

```python
sns.set_style("darkgrid")
plt.figure(figsize=(12, 6))
sns.boxplot(x='genre', y='budget', data=df, palette='crest', hue='genre')
plt.xticks(rotation=45, color='black')
plt.yticks(color='black')
plt.title('Budget Distribution by Genre', fontsize=14, weight='bold')
plt.xlabel('Genre', color='black')
plt.ylabel('Budget (in millions)', color='black')
plt.tight_layout()
plt.show()
plt.savefig("Boxplot.png", dpi=300, bbox_inches='tight')
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='rating', y='gross', size='votes', hue='genre', alpha=0.7, palette='magma')
plt.title('Rating vs Revenue (Bubble Size = votes)')
plt.xlabel('IMDb Rating')
plt.ylabel('Revenue')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
plt.savefig("ratingVSrevenue.png", dpi=300, bbox_inches='tight')
```

```python
avg_profit_by_year = df.groupby('year')['profit'].mean().reset_index()
plt.figure(figsize=(10,5))
sns.lineplot(data=avg_profit_by_year, x='year', y='profit', marker='o', color='gold')
plt.title('Average Profit Over Years', fontsize=14, weight='bold')
plt.xlabel('Year')
plt.ylabel('Average Profit')
plt.tight_layout()
plt.show()
plt.savefig("Averageprofit.png", dpi=300, bbox_inches='tight')
```

```python
top_directors = df.groupby('director')['profit'].sum().nlargest(10).reset_index()

plt.figure(figsize=(10,5))
ax = sns.barplot(x='profit', y='director', data=top_directors, palette='flare')
plt.title("Top 10 Directors by Total Box Office Profit", fontsize=14)
plt.xlabel("Total Profit (in billions)")
plt.ylabel("Director")
for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', label_type='center', padding=3, fontsize=9, color='black')
plt.tight_layout()
plt.show()
plt.savefig("top10 directors.png", dpi=300, bbox_inches='tight')
```

```python
yearly = df.groupby('year')['profit%'].mean().reset_index()
sns.lineplot(data=yearly, x='year', y='profit%', marker='o', color='teal')
plt.title("Average Profit % by Year")
plt.show()
```

```python
import squarify
top10 = df.sort_values(by='gross', ascending=False).head(10)
labels = [f"{name}\n${gross/1e9:.2f}B" for name, 
          gross in zip(top10['name'], 
                       top10['gross'])]
plt.figure(figsize=(12,7))
squarify.plot(
    sizes=top10['gross'], 
    label=labels, 
    alpha=0.8, 
    color=sns.color_palette('Spectral', 10)
)
plt.title('Top 10 Movies by Gross Revenue (Treemap)', 
          fontsize=14, weight='bold')
plt.axis('off')
plt.show()
plt.savefig("top10movies.png", dpi=300, bbox_inches='tight')
```

```python
# Group by rating and sum total gross
rating_gross = df.groupby('rating')['gross'].sum().sort_values(ascending=False).head(4)

# Plot pie chart
plt.figure(figsize=(8,8))
plt.pie(
    rating_gross, 
    labels=rating_gross.index, 
    autopct='%1.1f%%', 
    startangle=120, 
    colors=plt.cm.Set3.colors
)
plt.title('Total Revenue by Movie Rating', fontsize=14, weight='bold')
plt.show()
```

```python
company_profit = df.groupby('company')['profit'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,5))
ax = sns.barplot(
    y='company', 
    x='Total Profit', 
    data=company_profit, 
    palette='crest'
)
plt.title('Top 10 Production Companies by Total Profit', 
          fontsize=14, weight='bold')
plt.xlabel('Total Profit ($)')
plt.ylabel('Company')

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', label_type='center', 
                 padding=3, fontsize=9, color='black')
    
plt.tight_layout()
plt.show()
```
