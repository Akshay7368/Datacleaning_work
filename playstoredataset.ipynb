```python
plt.figure(figsize=(15,15))
plt.title('Analyzing over the data')
for i in range(0, len(numerical_features)):
    plt.subplot(3, 3, i+1)
    sns.kdeplot(x=numerical_features[i],data=df, fill=True, color='r')##kernal density estimator
    plt.xlabel(numerical_features[i])
    plt.tight_layout()
```

```python
plt.figure(figsize=(20,15))
category = ['Type', 'Content Rating']
for i in range(0, len(category)):
    plt.subplot(2, 2, i+1 )
    sns.countplot(x= category[i],data=df, palette='Set2', legend=False, hue=category[i])
    plt.xlabel(category[i])
    plt.xticks(rotation=45)
    plt.tight_layout()
```

```python
def handle_Size(size):
    
    if isinstance(size, str):
        # size is a string
        
        if 'M' in size or 'm' in size:
            # M is present in the string
            return float(size.replace('M', '').strip()) * 1024  # Convert MB to KB
        elif 'K' in size or 'k' in size:
            # k is present in the string
            return float(size.replace('k', '').strip()) # Already in KB
    
    return np.nan # Return NaN for any other cases
            
            
 
df_cleaned['Size'] = df_cleaned['Size'].apply(handle_Size)
```
